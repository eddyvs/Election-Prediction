```python

from datascience import *
import numpy as np

#polling data
polling_data='president_polls.csv'

data=Table().read_table(polling_data,encoding='utf-8')

#cleaning up extra junk
data=data.select('pct','state','candidate_name','start_date','end_date')
#cleaning up polling without state
data=data.where('state', are.not_equal_to('nan'))
data=data.sort('state',descending=False)


#split into the two candidates we want, K = Kamala T=trump
dataK=data.where('candidate_name', are.equal_to('Kamala Harris'))
dataT = data.where('candidate_name',are.equal_to('Donald Trump'))

#checking how many states are actually polled for Kamala Harris its only 44!!!
states1=make_array()
for stat in dataT.column('state'):
    if stat not in states1:
       states1=np.append(states1,stat)
print(len(states1))
del states1
```
